```python
def scrape_source(source):
    rows = []

    for url in source["urls"]:
        try:
            print("Scraping:", url)

            res = requests.get(
                url,
                headers=headers,
                timeout=20
            )
            res.raise_for_status()

            soup = BeautifulSoup(res.content, "lxml")

            # Get title
            title_tag = soup.find("h1")

            if title_tag:
                title = title_tag.get_text(" ", strip=True)
            else:
                title = soup.title.get_text(" ", strip=True)

            # Try common article containers
            block = None

            selectors = [
                ("article", None),
                ("main", None),
                ("div", "article-body"),
                ("div", "post-content"),
                ("div", "entry-content")
            ]

            for tag, cls in selectors:
                if cls:
                    candidate = soup.find(tag, class_=cls)
                else:
                    candidate = soup.find(tag)

                if candidate:
                    text = candidate.get_text(" ", strip=True)

                    if len(text) > 300:
                        block = candidate
                        break

            if block is None:
                print("[SKIPPED] Could not find article body")
                continue

            paragraphs = block.find_all("p")

            if paragraphs:
                text = " ".join(
                    p.get_text(" ", strip=True)
                    for p in paragraphs
                )
            else:
                text = block.get_text(" ", strip=True)

            if len(text) > 100:
                rows.append({
                    "source": source["name"],
                    "url": url,
                    "title": title,
                    "text": text
                })
            else:
                print("[SKIPPED] Article text too short")

            time.sleep(1)

        except Exception as e:
            print(f"[ERROR] {url} -> {e}")

    return pd.DataFrame(rows)
```

```python
corpus_a = scrape_source(SOURCE_A)
corpus_b = scrape_source(SOURCE_B)
```

```python
import requests
import time
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import matplotlib.pyplot as plt
from textblob import TextBlob
import spacy
from scipy.stats import ttest_ind

nlp = spacy.load("en_core_web_sm")

print("All packages and spaCy model loaded successfully!")
```

```python
corpus_a = scrape_source(SOURCE_A)
corpus_b = scrape_source(SOURCE_B)

corpus = pd.concat([corpus_a, corpus_b], ignore_index=True)

print("Total articles scraped:", len(corpus))
print(corpus[["source", "title"]])
```

```python
corpus = pd.concat(
    [corpus_a, corpus_b],
    ignore_index=True
)

print("Total articles scraped:", len(corpus))
```

```python
print(corpus[["source", "title"]].to_string(index=False))
```

```python
corpus.to_csv("two_source_corpus.csv", index=False)

print("CSV saved successfully.")
```

```python
def sentiment_polarity(text):
    return TextBlob(text).sentiment.polarity
```

```python
corpus["sentiment_polarity"] = corpus["text"].apply(
    sentiment_polarity
)
```

```python
def noun_density(text):
    doc = nlp(text)

    words = [
        token for token in doc
        if token.is_alpha
    ]

    nouns = [
        token for token in words
        if token.pos_ in {"NOUN", "PROPN"}
    ]

    if len(words) == 0:
        return 0

    return len(nouns) / len(words) * 100
```

```python
corpus["noun_density"] = corpus["text"].apply(
    noun_density
)
```

```python
corpus[
    [
        "source",
        "title",
        "sentiment_polarity",
        "noun_density"
    ]
].head(20)
```

```python
corpus.to_csv("two_source_corpus.csv", index=False)

print("Final CSV saved.")
```

```python
means = corpus.groupby("source")[
    ["sentiment_polarity", "noun_density"]
].mean()

means
```

```python
google_sentiment = means.loc[
    "google-blog",
    "sentiment_polarity"
]

python_sentiment = means.loc[
    "python-insider",
    "sentiment_polarity"
]

google_noun = means.loc[
    "google-blog",
    "noun_density"
]

python_noun = means.loc[
    "python-insider",
    "noun_density"
]

sentiment_difference = (
    google_sentiment - python_sentiment
)

noun_difference = (
    google_noun - python_noun
)

print(
    f"Mean sentiment difference: "
    f"{sentiment_difference:.4f}"
)

print(
    f"Mean noun density difference: "
    f"{noun_difference:.2f} percentage points"
)
```

```python
google_sentiment_values = corpus.loc[
    corpus["source"] == "google-blog",
    "sentiment_polarity"
]

python_sentiment_values = corpus.loc[
    corpus["source"] == "python-insider",
    "sentiment_polarity"
]

google_noun_values = corpus.loc[
    corpus["source"] == "google-blog",
    "noun_density"
]

python_noun_values = corpus.loc[
    corpus["source"] == "python-insider",
    "noun_density"
]
```

```python
sentiment_test = ttest_ind(
    google_sentiment_values,
    python_sentiment_values,
    equal_var=False
)

print("Sentiment t-test")
print("t-statistic:", sentiment_test.statistic)
print("p-value:", sentiment_test.pvalue)
```

```python
noun_test = ttest_ind(
    google_noun_values,
    python_noun_values,
    equal_var=False
)

print("Noun density t-test")
print("t-statistic:", noun_test.statistic)
print("p-value:", noun_test.pvalue)
```

```python
labels = ["Sentiment Polarity", "Noun Density"]

google_values = [
    google_sentiment,
    google_noun
]

python_values = [
    python_sentiment,
    python_noun
]

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 6))

ax.bar(
    x - width / 2,
    google_values,
    width,
    label="Google Blog"
)

ax.bar(
    x + width / 2,
    python_values,
    width,
    label="Python Insider"
)

ax.set_ylabel("Mean Value")
ax.set_title(
    "Mean Sentiment Polarity and Noun Density by Source"
)

ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

plt.tight_layout()
plt.savefig(
    "grouped_bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
sentiment_p = sentiment_test.pvalue
noun_p = noun_test.pvalue

report = f"""
BLOG SCRAPING COMPARISON REPORT

Sources:
1. Google Blog
2. Python Insider

Sample:
10 articles from Google Blog
10 articles from Python Insider
Total = {len(corpus)} articles

MEAN RESULTS

Google Blog mean sentiment polarity:
{google_sentiment:.4f}

Python Insider mean sentiment polarity:
{python_sentiment:.4f}

Difference in mean sentiment polarity:
{sentiment_difference:.4f}

Google Blog mean noun density:
{google_noun:.2f}%

Python Insider mean noun density:
{python_noun:.2f}%

Difference in mean noun density:
{noun_difference:.2f} percentage points

STATISTICAL TESTS

Sentiment polarity:
t-statistic = {sentiment_test.statistic:.4f}
p-value = {sentiment_p:.4f}

Noun density:
t-statistic = {noun_test.statistic:.4f}
p-value = {noun_p:.4f}

INTERPRETATION

The analysis compares ten articles from each of two different blog sources.
The observed difference in mean sentiment polarity is {sentiment_difference:.4f},
while the observed difference in mean noun density is {noun_difference:.2f}
percentage points. These differences describe the sample, but their practical
meaning should be interpreted carefully because the sample contains only ten
articles per source. With such a small sample, individual articles can have a
large effect on the group mean, and the selected articles may not represent all
articles published by either blog. Therefore, even a noticeable numerical
difference may not generalize to the wider populations of articles without a
larger and more systematically selected sample. The p-values from the
statistical tests should also be considered alongside the effect sizes and the
limitations of the small sample.
"""

with open("report.txt", "w", encoding="utf-8") as f:
    f.write(report)

print(report)
```
